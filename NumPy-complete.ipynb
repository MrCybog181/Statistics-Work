# Handle missing value in Numpy

```python
sal = np.array([12000,15000,np.nan,10000])
print(sal)
```

```python
sal = np.array([12000,15000,11000,10000])
print(sal)
```

```python
sal[0]=22000
print(sal)
```

```python
sals = np.array([12000,15000,11000,10000])
print(sal)
```

```python
sals[1]=np.int32(np.nan)
print(sals)
```

```python
# how to chnage datatype of array values after creation
# new_sals=np.astype(sals,'float')

new_sals=np.astype(sals,'float')
print(new_sals)
```

```python
new_sals[1]=np.nan
print(new_sals)
```

```python
#infinite in numpy
new_sals[0]=np.inf
new_sals
```

```python
# make new matrix to combine two matrix
a = np.array([[1,2],[3,4]])
b = np.array([[1,1],[2,2]])
print(a)
print(b)
```

```python
np.concatenate((a,b),axis=0)  #row concatenate(vertical stacking)
```

```python
np.concatenate((a,b),axis=1)  #col concatenate(horizontal stacking)
```

```python
d=np.concatenate((a,b),axis=None) #first flatten then concatenate
```
